## Manipulating tensors (tensor operations)

```python
# Create a tensor
tensor = torch.tensor([1, 2, 3])

# Adding
print(f"Adding: {tensor + 10}")
print("=="*30)

# Multiply
print(f"Multiply: {tensor * 10}")
print("=="*30)

# Element-wise multiplication
print(tensor, "*", tensor)
print("Equals:", tensor * tensor)
```

```python
# Matrix Mulitplication - Dot Product
print(f"Dot Product: {tensor @ tensor}")

#================================================
# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

print(f"Matrix Multiplication:")
print(tensor_A @ tensor_B.T)
```

```python
# Transpose 
"""
tensor.transpose(dim0, dim1) - tensor.T
"""
tensor_B.transpose(0, 1)
```

## Reshaping, stacking, squeezing and unsqueezing

```python
# Reshape the tensor
x = torch.arange(1., 8.)
x_reshaped = x.reshape(1, 7)


print(f"x: {x}, x.shape: {x.shape}")
print("==" * 30)
print(f"x reshaped: {x_reshaped}, x_reshaped.shape: {x_reshaped.shape}")
```

```python
# Change view (keeps same data as original but changes view)
z = x.view(1, 7)
print(f"z: {z}, z.shape: {z.shape}")
print("=="*30)

# Changing z changes x
z[:, 0] = 5
print(z, x)
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked
```

```python
# squeeze the tensor
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimension from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# Create tensor with specific shape
x_original = torch.rand(size=(224, 224, 3))

# Permute the original tensor to rearrange the axis order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

## Running tensors on GPUs

```python
# Check for GPU
torch.cuda.is_available()
```

```python
# Set device type
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
# Count number of devices for GPUs
torch.cuda.device_count()
```

```python
# Create tensor (default on CPU)
tensor = torch.tensor([1, 2, 3])

# Tensor not on GPU
print(tensor, tensor.device)

# Move tensor to GPU (if available)
tensor_on_gpu = tensor.to(device)
tensor_on_gpu

"""tensor([1, 2, 3], device='cuda:0')"""
# This must be the output if you have a GPU
```

```python
# copy the tensor back to cpu
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu
```
