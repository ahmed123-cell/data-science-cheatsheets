```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Get some data sets from seaborn
print(sns.get_dataset_names())

# Load the 3 most popular seaborn datasets
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
penguins = sns.load_dataset("penguins")
```

## Line Plot

```python
"""
sns.lineplot(
    data:                   # DataFrame containing the data to plot.
    x:                      # Column name for the x-axis.
    y:                      # Column name for the y-axis.
    estimator:              # How to aggregate multiple y values per x (e.g. 'mean', 'median').
    marker:                 # Marker shape for each data point (e.g. 'o' = circle, 's' = square).
    markersize:              # Size of each marker.
    color:                    # Color of the line.
    linewidth:                 # Thickness of the line.
    alpha:                      # Transparency of the line/markers (0 = fully transparent, 1 = fully opaque).
)
"""

# ===== Line Plot ===== #
plt.figure(figsize=(11, 4))

sns.lineplot(data=tips, x="day", y="total_bill", estimator="mean", marker="o", color='#e74c3c')

plt.title("Average Total Bill by Day of the Week", fontsize=14, fontweight='bold')
plt.xlabel("Day")
plt.ylabel("Average Total Bill ($)")
plt.grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

## Bar Plot

```python
"""
sns.barplot(
    x:                      # Column name for the x-axis (categories).
    y:                      # Column name for the y-axis (values).
    data:                   # DataFrame containing the data to plot.
    hue:                    # Column name used to split/color bars into subgroups.
    palette:                # Color palette used for the bars (or hue groups).
    order:                  # Custom order for the x-axis categories.
    estimator:              # How to aggregate multiple values per category (e.g. 'mean', 'median').
)
"""

# ===== Bar Plot ===== #
plt.figure(figsize=(11, 5))

sns.barplot(x="day", y="total_bill", data=tips, hue="sex", palette="Set2", estimator="mean")

plt.title("Average Total Bill by Day and Gender", fontsize=14, fontweight='bold')
plt.xlabel("Day of the Week")
plt.ylabel("Average Total Bill ($)")
plt.legend(title="Gender")
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Scatter Plot

```python
"""
sns.scatterplot(
    data:                   # DataFrame containing the data to plot.
    x:                      # Column name for the x-axis.
    y:                      # Column name for the y-axis.
    hue:                    # Column name used to color points by category.
    size:                   # Column name used to scale point size by value.
    style:                  # Column name used to vary marker shape by category.
    palette:                # Color palette used for the hue groups.
    alpha:                  # Transparency of the points (0 = fully transparent, 1 = fully opaque).
)
"""

# ===== Scatter Plot ===== #
plt.figure(figsize=(11, 5))

sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species", style="species", palette="Set2")

plt.title("Sepal Length vs Sepal Width by Species (Iris Dataset)", fontsize=14, fontweight='bold')
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.legend(fontsize=8, loc="upper left")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Histogram Plot

```python
"""
sns.histplot(
    data:                   # DataFrame containing the data to plot.
    x:                      # Column name for the x-axis.
    bins:                   # Number of bars the data is grouped into.
    kde:                    # If True, overlays a smooth density curve on top of the histogram.
    color:                  # Fill color of the bars.
    alpha:                  # Transparency of the bars (0 = fully transparent, 1 = fully opaque).
    stat:                   # What the bar heights represent (e.g. 'count', 'density', 'probability').
)
"""

# ===== Histogram ===== #
plt.figure(figsize=(11, 5))

sns.histplot(data=tips, x="total_bill", bins=25, kde=True, color="#1f77b4", alpha=0.7)

mean_val = tips['total_bill'].mean()
std_val = tips['total_bill'].std()

plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: ${mean_val:.2f}')
plt.axvline(mean_val + std_val, color='green', linestyle=':', linewidth=2, label=f'+1 Std: ${mean_val + std_val:.2f}')
plt.axvline(mean_val - std_val, color='green', linestyle=':', linewidth=2, label=f'-1 Std: ${mean_val - std_val:.2f}')

plt.title("Distribution of Total Bill Amounts (Tips Dataset)", fontsize=14, fontweight='bold')
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## KDE Plot

```python
"""
sns.kdeplot(
    data:                   # DataFrame containing the data to plot.
    x:                      # Column name for the x-axis.
    hue:                    # Column name used to split the density curve by category.
    fill:                   # If True, fills the area under the curve with color.
    alpha:                  # Transparency of the fill (0 = fully transparent, 1 = fully opaque).
    linewidth:               # Thickness of the density curve line.
    palette:                  # Color palette used for the hue groups.
    multiple:                  # How overlapping curves are displayed ('layer', 'stack', or 'dodge').
)
"""

# ===== KDE Plot ===== #
plt.figure(figsize=(11, 5))

sns.kdeplot(data=penguins, x="body_mass_g", hue="species", fill=True, alpha=0.6, palette="Set2")

plt.title("Distribution of Penguin Body Mass by Species", fontsize=14, fontweight='bold')
plt.xlabel("Body Mass (grams)")
plt.ylabel("Density")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Box Plot

```python
"""
sns.boxplot(
    x:                      # Column name for the x-axis (categories).
    y:                      # Column name for the y-axis (values).
    data:                   # DataFrame containing the data to plot.
    hue:                    # Column name used to split boxes into subgroups.
    palette:                # Color palette used for the hue groups.
    width:                  # Width of each box.
    showmeans:              # If True, also displays the mean of each group.
)
"""

# Enhanced Box Plot
plt.figure(figsize=(12, 5))

sns.boxplot(
    x="day",
    y="total_bill",
    data=tips,
    hue="sex",
    palette="Set2",
    order=["Thur", "Fri", "Sat", "Sun"],
    hue_order=["Male", "Female"], 
    width=0.7, 
    whis=1.5,
    showfliers=True, 
    linewidth=1.8, 
    fliersize=6,
    showmeans=True,
    
    # Custom properties
    whiskerprops=dict(color="black", linewidth=1.5),
    capprops=dict(color="black", linewidth=1.5),
    boxprops=dict(edgecolor="black", linewidth=1.5),
    medianprops=dict(color="#e74c3c", linewidth=3),
    meanprops=dict(marker='^', markerfacecolor='white', markeredgecolor='black'),
    
    orient="v"
)

plt.title("Total Bill Distribution by Day and Gender", fontsize=16, pad=15, fontweight='bold')
plt.xlabel("Day of the Week", fontsize=12)
plt.ylabel("Total Bill ($)", fontsize=12)

plt.legend(title="Gender", title_fontsize=12)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Violin Plot

```python
"""
sns.violinplot(
    x:                      # Column name for the x-axis (categories).
    y:                      # Column name for the y-axis (values).
    data:                   # DataFrame containing the data to plot.
    hue:                    # Column name used to split violins into subgroups.
    split:                  # If True, draws each hue pair as a single split violin instead of side-by-side.
    palette:                # Color palette used for the hue groups.
    inner:                  # What's shown inside the violin ('box', 'quartile', 'point', 'stick', or None).
)
"""

# Enhanced Violin Plot
plt.figure(figsize=(12, 5))

sns.violinplot(
    x="day",
    y="total_bill", 
    data=tips,
    hue="sex",
    split=True,                 # split violins by hue
    palette="Set1",
    order=["Thur", "Fri", "Sat", "Sun"],
    hue_order=["Male", "Female"],
    inner="quartile",           # inner: 'box', 'quartile', 'point', 'stick', or None
    linewidth=1.5,
    cut=0                       # limit the violin range
)

plt.title("Total Bill Distribution by Day and Gender (Violin Plot)", fontsize=16, pad=15, fontweight='bold')
plt.xlabel("Day of the Week", fontsize=12)
plt.ylabel("Total Bill ($)", fontsize=12)

plt.legend(title="Gender", title_fontsize=12, loc="upper right")
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Strip Plot: A strip plot displays individual data points as dots along a categorical axis. It uses small random jitter to reduce overlap, making it easy to see the distribution and density of values within each category.
---
## Swarm Plot: A swarm plot (also known as a beeswarm plot) is similar to a strip plot but intelligently arranges points to avoid any overlap. It provides a clearer view of the actual data distribution and density without jitter.

```python
# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(15, 5), dpi=100)

"""
sns.stripplot / sns.swarmplot(
    data:                   # DataFrame containing the data to plot.
    x:                      # Column name for the x-axis (categories).
    y:                      # Column name for the y-axis (values).
    hue:                    # Column name used to split/color points into subgroups.
    palette:                # Color palette used for the hue groups.
    size:                   # Size of each point.
    dodge:                  # If True, separates hue groups side-by-side instead of overlapping.
)
"""

# ==================== Strip Plot ====================
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="viridis",
    size=7,
    alpha=0.7,
    jitter=True,
    dodge=True,
    edgecolor="black",
    linewidth=0.5,
    ax=axs[0]
)
axs[0].set_title("Strip Plot - Total Bill by Day", fontsize=14, fontweight='bold')
axs[0].set_xlabel("Day")
axs[0].set_ylabel("Total Bill ($)")

# ==================== Swarm Plot ====================
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="flare",
    size=6,
    alpha=0.85,
    dodge=True,
    edgecolor="black",
    linewidth=0.5,
    ax=axs[1]
)
axs[1].set_title("Swarm Plot - Total Bill by Day", fontsize=14, fontweight='bold')
axs[1].set_xlabel("Day")
axs[1].set_ylabel("Total Bill ($)")

plt.suptitle("Strip Plot vs Swarm Plot (Tips Dataset)", fontsize=16, y=1.02, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Heat Map

```python
# Select only numerical columns and drop missing values & Calculate correlation matrix
numeric_penguins = penguins.select_dtypes(include=np.number).dropna()
corr_matrix = numeric_penguins.corr()

"""
sns.heatmap(
    corr_matrix,
    annot:                  # If True, writes the actual value inside each cell.
    cmap:                   # Colormap used to translate values into colors (e.g. 'Blues', 'viridis').
    center:                 # Value at which the colormap is centered (useful for correlation data around 0).
    square:                 # If True, forces each cell to be square-shaped.
    fmt:                    # Number format for the annotated values (e.g. '.2f' = 2 decimal places).
)
"""

# ===== Heatmap ===== #
plt.figure(figsize=(10, 5))

sns.heatmap(corr_matrix, annot=True, cmap="Blues", center=0, square=True, fmt=".2f")

plt.title("Correlation Heatmap of Penguin Measurements", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Pair Plot

```python
# Enhanced Pair Plot
sns.pairplot(
    data=iris,
    hue="species",
    kind="scatter",                   # kind: 'scatter', 'reg', or 'kde'
    diag_kind="kde",                  # diag_kind: 'hist' or 'kde'
    vars=["sepal_length", "sepal_width", "petal_length", "petal_width"],
    palette="magma",
    height=3.5,
    aspect=1.1,
    markers=["o", "s", "D"],
    plot_kws={"alpha": 0.8, "s": 50},
    diag_kws={"linewidth": 2, "alpha": 0.7}
)

plt.suptitle("Pair Plot of Iris Dataset", fontsize=18, y=1.02, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Pair Grid

```python
# Create PairGrid
g = sns.PairGrid(
    data=iris,
    hue="species", 
    vars=["sepal_length", "sepal_width", "petal_length", "petal_width"],  # variables to plot
    palette="Set2",
    height=3.5,
    aspect=1.1,
    corner=True,                # Only show lower triangle (widely used)
    diag_sharey=False
)

# Map plots to the grid
g.map_diag(sns.histplot, kde=True, alpha=0.7)                          # Diagonal: Histogram + KDE
g.map_offdiag(sns.scatterplot, alpha=0.8, size=40, edgecolor="black")  # Off-diagonal: Scatter

# Add legend
g.add_legend(title="Species", title_fontsize=12, fontsize=11)

g.figure.suptitle("PairGrid - Iris Dataset", fontsize=18, y=1.02)
plt.tight_layout()
plt.show()
```

## Facet Grid

```python
# Create FacetGrid
g = sns.FacetGrid(
    data=tips,
    col="time",
    row="sex",
    hue="smoker",
    palette="crest",
    height=4,
    aspect=1.2,
    margin_titles=True,   # Show titles on the margins
    col_wrap=None
)

# Map a scatter plot to the grid
g.map(sns.scatterplot, "total_bill", "tip", alpha=0.8, s=60, edgecolor="black")

# Add regression lines (optional but very useful)
g.map(sns.regplot, "total_bill", "tip", scatter=False, line_kws={"linewidth": 2.5})

# Add legend and titles
g.add_legend(title="Smoker", title_fontsize=12)
g.set_axis_labels("Total Bill ($)", "Tip ($)")
g.set_titles(row_template="{row_name}", col_template="{col_name}")

g.figure.suptitle("FacetGrid: Total Bill vs Tip by Time, Sex & Smoking Status", 
               fontsize=16, y=1.02)

plt.tight_layout()
plt.show()
```
