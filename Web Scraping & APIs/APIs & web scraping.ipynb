## Beautiful Soup (bs4)

```python
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")
```

```python
# find(name, attrs, class_, id): find the first element

soup.find('h3')
soup.find("div", class_="product_price") # --> class
soup.find(id='main') # --> ID
```

```python
#find_all(name, attrs, class_, id, limit)
soup.find_all("a")
```

```python
# Select: Uses CSS selector

soup.select('h3')
soup.select('.price_color')     # --> class
soup.select('#main')            # --> id
soup.select("article h3 a")     # Tag inside a Tag
```

```python
# text
book = soup.find("h3")
print(book.text)
```

```python
book.attrs        # --> Access Attributes
book.parent       # --> parent
book.children     # --> child
book.next_sibling # --> the next element 
```

## Practice Example

```python
# =============== Extract Books Titles =============== #
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

books = soup.find_all("h3")

for book in books:
    print(book.a["title"])
```

```python
# =============== Extract Books Prices =============== #
prices = soup.find_all("p", class_="price_color")

for price in prices:
    print(price.get_text())
```

```python
# =============== Extract Books Links =============== #
links = soup.find_all("a")

for link in links:
    print(link.get("href"))
```
